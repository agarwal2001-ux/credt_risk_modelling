```python
import pandas as pd
import numpy as np
from scipy.stats import chi2_contingency,f_oneway
from statsmodels.stats.outliers_influence import variance_inflation_factor
```

```python
df = pd.read_csv("modeldf.csv")
```

```python
numcols = df.select_dtypes(include=["int64","float64"]).columns.tolist()
numcols.remove("prospectid")
catcols = df.select_dtypes(include=["object","category"]).columns.tolist()
```

```python
numdf = df[numcols]
```

```python

columns_to_be_kept_numerical = []
col_name = []
p_val = []

for i in numcols:
    a = list(numdf[i])  
    b = list(df['approved_flag'])  
    
    group_P1 = [value for value, group in zip(a, b) if group == 'P1']
    group_P2 = [value for value, group in zip(a, b) if group == 'P2']
    group_P3 = [value for value, group in zip(a, b) if group == 'P3']
    group_P4 = [value for value, group in zip(a, b) if group == 'P4']


    f_statistic, p_value = f_oneway(group_P1, group_P2, group_P3, group_P4)

    if p_value <= 0.05:
        col_name.append(i)
        p_val.append(p_value)
        columns_to_be_kept_numerical.append(i)
```

```python
pd.DataFrame({"col_name":col_name,"p_value":p_val})
```

```python
after_anova = numdf[columns_to_be_kept_numerical]
```

```python
vif_data = numdf[columns_to_be_kept_numerical]
total_columns = vif_data.shape[1]
columns_to_be_kept = []
vif = []
column_index = 0

for i in range (0,total_columns):
    
    vif_value = variance_inflation_factor(vif_data, column_index)
    print (column_index,'---',vif_value)
    
    
    if vif_value <= 6:
        columns_to_be_kept.append( columns_to_be_kept_numerical[i] )
        vif.append(vif_value)
        column_index = column_index+1
    
    else:vif_data = vif_data.drop([ columns_to_be_kept_numerical[i] ] , axis=1)
```

```python
pd.DataFrame({"col_name":columns_to_be_kept,"vif_val":vif})
```
