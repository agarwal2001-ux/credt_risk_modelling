```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.read_csv("case_study1.csv")
df2 = pd.read_csv("case_study2.csv")
```

```python
df1 = df1.replace(-99999,np.nan)
```

```python
df2 = df2.replace(-99999,np.nan)
```

```python
df1.columns = df1.columns.str.lower()
df2.columns = df2.columns.str.lower()
```

```python
new_df = df1.merge(df2,left_on="prospectid",right_on="prospectid",how="inner")
```

```python
new_df.to_csv("merged.csv",index=True)
```
